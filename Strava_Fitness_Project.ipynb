# Strava Fitness & Wellness Analysis

### Data Analysis Project

This project analyzes user activity, sleep, and weight/BMI data to understand fitness behavior, engagement patterns, and activity trends.

## Business Problem

A fitness and wellness company collects data from users through fitness tracking devices. The data includes daily activity, sleep patterns, and weight/BMI information.

The company wants to understand how users engage with fitness tracking and identify patterns in their activity and wellness behaviour.

The analysis will help answer questions such as:

- How active are users on a typical day?
- How consistent is user activity?
- How do activity levels vary across days?
- What are the users' sleep patterns?
- What weight and BMI patterns can be observed among users who track their weight?
- What insights can help the company improve user engagement and encourage healthier, more consistent behaviour?

## Project Objectives

The main objectives of this analysis are to:

1. Analyze users' daily activity levels using steps, distance, calories, and activity minutes.
2. Identify patterns in user activity across different days of the week and over the tracking period.
3. Analyze sleep duration and time spent in bed.
4. Examine weight and BMI patterns among users who recorded weight data.
5. Identify differences in activity levels across users.
6. Generate data-driven insights that can help improve user engagement and support healthier, more consistent fitness behaviour.

## User Activity Variation

Average daily steps were calculated for each user to understand differences in activity levels.
